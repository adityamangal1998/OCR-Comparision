```python
import cv2
import pytesseract

# Path to the Tesseract executable (for Windows)
pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'
# Read image
img = cv2.imread('test_image.png')
# Convert to RGB (pytesseract prefers RGB)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# Perform OCR
text = pytesseract.image_to_string(img_rgb, lang='hin+eng')
print("-----Extracted Text-----")
print(text)
```

```python
custom_config = r'--oem 3 --psm 6'
text = pytesseract.image_to_string(img_rgb, lang='hin+eng', config=custom_config)
print("-----Extracted Text-----")
print(text)
```
